```python
from sklearn import preprocessing
```

```python
le = preprocessing.LabelEncoder()
data['diagnosis'] = le.fit_transform(data['diagnosis'])
```

```python
data.head()
```

```python
from sklearn.model_selection import train_test_split 
```

```python
Features = data.drop('diagnosis' , axis = 1)
Labels = data['diagnosis']
```

```python
Features.shape , Labels.shape
```

```python
x_train , x_test , y_train , y_test = train_test_split(Features , Labels , test_size = 0.2)
```

```python
x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
obj = RandomForestClassifier()
```

```python
obj.fit(x_train , y_train)
```

```python
preds = obj.predict(x_train)
preds.shape
```

```python
train_acc = accuracy_score(y_train , preds)
train_acc
```

```python
cm_train = confusion_matrix(y_train , preds)
cm_train
```

```python
preds2 = obj.predict(x_test)
preds2.shape
```

```python
test_acc = accuracy_score(y_test , preds2)
test_acc
```

```python
cm_test = confusion_matrix(y_test , preds2)
cm_test
```
